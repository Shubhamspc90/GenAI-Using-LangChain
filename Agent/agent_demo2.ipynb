```python
import os
from dotenv import load_dotenv

load_dotenv()

ollama_api_key = os.getenv("OLLAMA_API_KEY")

print("OLLAMA API KEY FOUND:", bool(ollama_api_key))
```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain.tools import tool
```

```python
# model
llm = ChatOllama(
    model="gpt-oss:120b",
    base_url="https://ollama.com",
    client_kwargs={
        "headers": {
            "Authorization": f"Bearer {ollama_api_key}"
        }
    },
    temperature=0
)
```

```python
# just to check model is working or not
llm.invoke('Hi ,how are you')
```

```python
# tool
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()

result = search_tool.invoke("which is the longest river of Dubai ?")
```

```python
# just to check the took ,is this working or not
result
```

```python
# abhi tk  hamare pass ( llm + tool ) aa chuka hai ,,ab Agent banane ki baari hai
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[search_tool],
    system_prompt="""
    You are a helpful assistant.
    Use the web search tool when the user asks for current or latest information.
    """
)
```

```python
# Ask the agent

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is the current time of Dubai and weather?"
        }
    ]
})
```

```python
# Result 
print(response["messages"][-1].content)
```

```python
# User
#   ↓
# agent.invoke()
#   ↓
# create_agent
#   ↓
# Ollama Cloud (gpt-oss:120b)
#   ↓
# LLM decides:
# "I need current information → use search_tool"
#   ↓
# DuckDuckGoSearchRun
#   ↓
# Search results
#   ↓
# Results sent back to LLM
#   ↓
# Ollama Cloud
#   ↓
# Final answer
```

```python
# User Query → Agent ko Dubai ka current time aur weather poocha.
# Agent Analysis → Agent ne identify kiya ki ye current information hai.
# Tool Selection → Agent ne DuckDuckGo Search Tool use kiya.
# Search → Dubai ke current time aur weather ki information search ki.
# LLM → Search results ko samajhkar final answer banaya.
# Response → Final answer response["messages"][-1].content se nikala.
```
