```python
# loading API key
import os
from dotenv import load_dotenv
load_dotenv()
ollama_api_key = os.getenv("OLLAMA_API_KEY")
print("ollama api key found : ",bool(ollama_api_key))
```

```python
# ============================================================
# 1. MODEL CREATION
# ============================================================

from langchain_ollama import ChatOllama
llm = ChatOllama(
    model = "gpt-oss:20b-cloud",
    base_url = "https://ollama.com",
    client_kwargs = {
        "headers":{
            "Authorization" : f"Bearer {ollama_api_key}"
        }
    },
    temperature = 0.1
)

```

```python
llm.invoke("hi")
```

```python
# ============================================================
# 2. TOOL CREATION
# ============================================================
# @tool -> Python function ko LangChain Tool banata hai.
from langchain_core.tools import  tool

@tool 

def calculate_gst(amount:float , gst_rate:float ) -> dict :
    """ Calculte the GST and Total amount"""

    gst = amount * gst_rate / 100
    total = amount + gst
    return{
        "Amount ": amount,
        "gst_rate " : gst_rate,
        "GST " : gst,
        "Total " : total    
    }


```

```python
# ============================================================
# 3. AGENT CREATION
# ============================================================
# Yahan hum LLM + Tool ko Agent ke andar de rahe hain.
#
# Agent ko pata hai:
#   - kaunsa model use karna hai
#   - kaunse tools available hain
#
# IMPORTANT:
# create_agent() ke andar tool automatically
# model ke saath connect/bind ho jata hai.

from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[calculate_gst]
)
```

```python
# ============================================================
# 4. USER INPUT
# ============================================================

user_question = "Calculate 18% GST on 5000" 
```

```python
# ============================================================
# 5. AGENT INVOCATION
# ============================================================
# Yahan actual Agent start hota hai.
#
# Agent internally:
#
# User Question
#      ↓
# LLM
#      ↓
# TOOL CALLING
#      ↓
# Tool select
#      ↓
# TOOL EXECUTION
#      ↓
# Tool Result
#      ↓
# LLM
#      ↓
# Final Answer

result = agent.invoke({
    "messages" : [
        {
            "role" : "user",
            "content" : user_question
        }
    ]
})
```

```python
# ============================================================
# 6. FINAL RESPONSE
# ============================================================

print(result["messages"][-1].content)
```
