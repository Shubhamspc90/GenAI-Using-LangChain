```python
# import libraries
from google.colab import userdata
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain.agents import create_agent
from ddgs import DDGS

ollama_api_key = userdata.get("OLLAMA_API_KEY")
print("ollama key Found : ", bool(ollama_api_key))
```

```python
# model creation
llm = ChatOllama(
    model="gpt-oss:20b-cloud",
    base_url="https://ollama.com",
    client_kwargs={
        "headers": {
            "Authorization": f"Bearer {ollama_api_key}"
        }
    },
    temperature=0.1
)
```

```python
#  checking model working
print(llm.invoke("hi").content)
```

```python
# tool creation
@tool
def search_web(query: str) -> list:
    """Search the web and return search results."""
    search = DDGS()
    results = search.text(query, max_results=5)
    return results
```

```python
# checking tool working
result = search_web.invoke({
    "query": "What is LangChain?"
})

print(result)
```

```python
# creating Agent
agent = create_agent(
    model = llm,
    tools = [search_web]
)
```

```python
user_question1 = "What is the latest version of LangChain?"

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": user_question1
        }
    ]
})

print(result["messages"][-1].content)
```

```python
user_question2 = "What are the latest developments in Generative AI?"

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": user_question2
        }
    ]
})

print(result["messages"][-1].content)
```
